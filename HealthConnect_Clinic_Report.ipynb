# 1. Problem Definition

## 1.1 Background

HealthConnect Clinic is a fictional healthcare provider that manages appointment-based services. The clinic is experiencing challenges related to patients missing scheduled appointments, difficulty understanding factors associated with appointment no-shows, and inefficient use of appointment slots when patients fail to attend. The clinic therefore wants to explore how data, machine learning, and Generative AI can support better decision-making and improve the patient experience.

## 1.2 Data Science Problem

From the Data Science perspective, the main problem is to determine whether the available appointment data can support the development of a machine-learning solution for predicting patient appointment no-shows.

The problem will be treated as a **supervised binary classification problem**. The proposed model will learn from historical appointment records and identify patterns that may be associated with whether a patient attends or misses a scheduled appointment.

The Data Science track is specifically responsible for defining the machine-learning problem and assessing whether the available data can support a no-show prediction solution.

## 1.3 Proposed Machine Learning Objective

The proposed objective is:

> **To assess whether the HealthConnect appointment dataset can be used to develop a machine-learning model that predicts whether a scheduled patient appointment will result in a no-show.**

The model will use relevant information available in the appointment dataset, such as patient characteristics, appointment details, booking information, previous appointment history, reminder information, distance to the clinic, and other suitable variables.

## 1.4 Proposed Target Variable

The dataset contains `appointment_outcome` as the final appointment status, with the categories:

* Attended
* No-Show
* Cancelled

For the initial prediction problem, the proposed target variable is **No-Show**.

A binary target can be derived as follows:

| Appointment Outcome |    Proposed Target |
| ------------------- | -----------------: |
| Attended            |                  0 |
| No-Show             |                  1 |
| Cancelled           | Excluded initially |

Cancelled appointments will initially be treated separately from no-shows because a cancellation represents a different appointment outcome from a patient simply failing to attend.

The handling of cancellations will be reviewed further during the modelling stage.

## 1.5 Expected Machine Learning Input and Output

The proposed machine-learning process is:

**Appointment and historical information → Machine-learning model → No-show prediction**

Potential model input variables include:

* Age
* Gender

* Appointment time
* Booking lead days
* Previous appointments
* Previous no-shows
* Reminder sent

These variables will not automatically be included in the final model. Their suitability will first be assessed during the initial data assessment to identify missing values, irrelevant variables, possible redundancy, and potential data leakage.

## 1.6 Expected Project Contribution

The proposed prediction solution could help HealthConnect identify appointments that have a higher likelihood of becoming no-shows. This information could eventually support more targeted appointment management and patient engagement activities.

However, at this stage, the project does not assume that a model will automatically reduce no-shows. Week 4 is focused on understanding the problem, assessing the available resources, defining the approach, and preparing for subsequent development.

The initial scope of the Data Science work is limited to **predicting appointment no-shows using the provided appointment data**. The work will focus on assessing the data, defining the target and potential features, and proposing an appropriate modelling approach.


# 2. Initial Data Assessment

## 2.1 Dataset Overview

The HealthConnect appointment dataset contains fictional and anonymised appointment records. The dataset provides information about patient demographics, appointment details, booking information, previous appointment history, reminder information, distance to the clinic, waiting time, and appointment outcomes. These variables provide a basis for assessing whether the data can support the proposed no-show prediction problem.

The dataset contains an `appointment_id` that serves as the unique identifier for each appointment and a `patient_id` representing an anonymised patient identifier. Since a patient may have multiple appointments, `patient_id` can appear across multiple records.

The available variables can broadly be grouped into the following categories:

| Variable Category                 | Examples                                                                      |
| --------------------------------- | ----------------------------------------------------------------------------- |
| Patient information               | `gender`, `age`, `age_group`                                                  |
| Appointment information           | `appointment_type`, `appointment_date`, `appointment_day`, `appointment_time` |
| Booking information               | `booking_date`, `booking_lead_days`                                           |
| Appointment history               | `previous_appointments`, `previous_no_shows`                                  |
| Reminder information              | `reminder_sent`, `reminder_channel`                                           |
| Accessibility and service factors | `distance_to_clinic_km`, `waiting_time_minutes`                               |
| Outcome information               | `appointment_outcome`                                                         |

The data dictionary identifies `appointment_outcome` as the final appointment status, with three categories: **Attended, No-Show, and Cancelled**.

## 2.2 Data Suitability for the Proposed Machine Learning Problem

The dataset appears relevant to the proposed no-show prediction problem because it contains several variables that may provide information about factors associated with appointment attendance.

For example, `previous_no_shows` may provide information about a patient's previous attendance behaviour, while `booking_lead_days` may indicate how far in advance an appointment was booked. Variables such as `reminder_sent`, `reminder_channel`, `distance_to_clinic_km`, appointment timing, and appointment type may also provide useful information for modelling.

However, the suitability of these variables cannot be confirmed only from the data dictionary. Further examination of the actual dataset will be required to determine the quality, distribution, completeness, and consistency of the data.

## 2.3 Data Quality Considerations

Several potential data-quality issues are identified from the data dictionary.

### Missing Values

`distance_to_clinic_km` and `waiting_time_minutes` are reported to contain limited missing values. These missing observations will need to be investigated before modelling.

The proportion and pattern of missing values should be assessed to determine whether the affected records can be retained through an appropriate imputation method or whether another treatment is required.

### Data Consistency

The relationship between historical appointment variables should also be checked. In particular, `previous_no_shows` should not exceed `previous_appointments`, as specified in the data dictionary.

The values of categorical variables such as `gender`, `appointment_type`, `reminder_sent`, `reminder_channel`, and `appointment_outcome` should also be checked for unexpected or inconsistent categories.

### Derived Variables

Some variables are already derived from other variables. For example, `age_group` is derived from `age`, while `appointment_day` is derived from `appointment_date`. `booking_lead_days` is also described as both provided and derivable.

These relationships should be considered during preprocessing to avoid unnecessary redundancy in the final feature set.

### Identifier Variables

`appointment_id` and `patient_id` should not automatically be treated as predictive features.

`appointment_id` is a unique identifier and is unlikely to contain meaningful predictive information. `patient_id` is anonymised and may occur across multiple appointments. Its use therefore requires careful consideration to prevent the model from learning patient-specific identifiers rather than general patterns.

## 2.4 Potential Data Leakage

Data leakage is an important consideration for the proposed prediction model.

The model should only use information that would realistically be available **before the prediction is made**. Variables that become known after the appointment or after the outcome has occurred should not be used as predictive features.

For example, `appointment_outcome` cannot be used as an input feature because it is the variable from which the proposed no-show target will be derived.

Similarly, `waiting_time_minutes` requires further investigation. If the estimated waiting time is available before the appointment and can reasonably be used at prediction time, it may be considered as a feature. If it is only known after the appointment process has started, including it could introduce data leakage.

The proposed target will be derived from `appointment_outcome`.
